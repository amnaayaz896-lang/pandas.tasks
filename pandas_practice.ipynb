## Task 5

Create a DataFrame containing some missing marks.

For example:

```
Ali       80
Sara      NaN
Ahmed     70
Ayesha    NaN
Zain      90
```
Find:

- Number of missing values
- Fill missing values with average marks
- Display the cleaned DataFrame

````python
import pandas as pd
````

````python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45],
}
df = pd.DataFrame(data)
print(df)
marks = df["Marks"]

print("\nAverage Marks")
print(marks.mean())

print("\nMaximum Marks")
print(marks.max())

print("\nMinimum Marks")
print(marks.min())

name = df["Name"]

print("\nTotal Students")
print(name.count())
````

````python
good_students =   marks[marks > 50]
print("Good Students")
print(good_students)

bad_students = marks[marks < 50]
print("\nBad Students")
print(bad_students)
````

````python
df["City"] = ["Karachi", "Lahore", "Islamabad", "Rawalpindi", "Peshawar"]
print(df)

print("Karachi Student")
print(df.loc[df["City"] == "Karachi"])
````

````python
data = {
    "Name": ["Ayesha", "Amna", "Maheen", "Usama", "Umer"],
    "Course": ["AI", "Web", "AI", "Web", "AI"],
    "Marks": [56, 78, 48, 39, 85],
}

df = pd.DataFrame(data)

result = df.groupby("Course")["Marks"].mean()
print(result)
````

````python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [80, np.nan, 70, np.nan, 90]
}

df = pd.DataFrame(data)
print(df)

print("\nMissing Values")
print(df.isnull().sum())

print("\nAverage Marks")
print(df["Marks"].mean())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print("\nCleaned Data")
print(df)

````
